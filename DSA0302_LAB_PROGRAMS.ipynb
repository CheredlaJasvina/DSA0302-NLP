```python

import re

text = "Python is a programming language. Python is easy to learn."

# Match a pattern at the beginning
result1 = re.match("Python", text)
print("Match:", result1.group() if result1 else "Not found")

# Search for a pattern anywhere
result2 = re.search("programming", text)
print("Search:", result2.group() if result2 else "Not found")

# Find all occurrences of a pattern
result3 = re.findall("Python", text)
print("Find All:", result3)

# Replace a pattern
result4 = re.sub("Python", "Java", text)
print("Replace:", result4)
```

```python

def finite_automaton(string):
    state = 0

    for char in string:
        if state == 0:
            if char == 'a':
                state = 1
            else:
                state = 0

        elif state == 1:
            if char == 'a':
                state = 1
            elif char == 'b':
                state = 2

        elif state == 2:
            if char == 'a':
                state = 1
            else:
                state = 0

    if state == 2:
        return "Accepted"
    else:
        return "Rejected"


string = input("Enter a string: ")
print("Result:", finite_automaton(string))
```

```python

import nltk
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

# Initialize stemmer and lemmatizer
ps = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = ["playing", "played", "studies", "running", "better"]

print("Word\t\tStemming\tLemmatization")

for word in words:
    stem = ps.stem(word)
    lemma = lemmatizer.lemmatize(word, pos='v')
    print(word, "\t\t", stem, "\t\t", lemma)
```

```python

def plural_form(noun):
    if noun.endswith(('s', 'sh', 'ch', 'x', 'z')):
        return noun + "es"
    elif noun.endswith('y') and noun[-2] not in 'aeiou':
        return noun[:-1] + "ies"
    else:
        return noun + "s"


# Input noun
word = input("Enter a noun: ")

# Generate plural form
plural = plural_form(word)

print("Singular Noun:", word)
print("Plural Noun:", plural)
```

```python

import random

# Sample training text
text = "I love python. I love programming. Python is easy."

# Split text into words
words = text.split()

# Create bigram model
bigram = {}

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]

    if current_word not in bigram:
        bigram[current_word] = []

    bigram[current_word].append(next_word)

# Generate text
word = "I"
generated_text = [word]

for i in range(5):
    if word in bigram:
        word = random.choice(bigram[word])
        generated_text.append(word)
    else:
        break

print("Generated Text:", " ".join(generated_text))
```

```python

import nltk
from nltk import RegexpTagger, word_tokenize

text = "The cat is playing"

patterns = [
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*s$', 'NNS'),
    (r'The|the|A|a', 'DT'),
    (r'is|are|am', 'VB'),
    (r'.*', 'NN')
]

tagger = RegexpTagger(patterns)

words = word_tokenize(text)
result = tagger.tag(words)

print("POS Tagging:")
print(result)
```

```python

words = ["playing", "walked", "cats", "book"]

# Initial tagging
tags = ["NN"] * len(words)

# Transformation rules
for i, word in enumerate(words):
    if word.endswith("ing"):
        tags[i] = "VBG"
    elif word.endswith("ed"):
        tags[i] = "VBD"
    elif word.endswith("s"):
        tags[i] = "NNS"

# Display result
for word, tag in zip(words, tags):
    print(word, ":", tag)
```

```python

import nltk

grammar = nltk.CFG.fromstring("""
S -> NP_S VP_S | NP_P VP_P
NP_S -> 'the' 'boy'
NP_P -> 'the' 'boys'
VP_S -> 'runs'
VP_P -> 'run'
""")

parser = nltk.ChartParser(grammar)

sentences = [
    "the boy runs",
    "the boys run",
    "the boy run"
]

for sentence in sentences:
    words = sentence.split()
    result = list(parser.parse(words))

    if result:
        print(sentence, ": Correct Agreement")
    else:
        print(sentence, ": Incorrect Agreement")
```

```python

import nltk

grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.8] | N [0.2]
VP -> V NP [0.7] | V [0.3]
Det -> 'the' [1.0]
N -> 'cat' [0.5] | 'fish' [0.5]
V -> 'eats' [1.0]
""")

parser = nltk.ViterbiParser(grammar)

sentence = "the cat eats fish".split()

for tree in parser.parse(sentence):
    print(tree)
    print("Probability:", tree.prob())
```

```python

import nltk
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')

# Input sentence
sentence = "I deposited money in the bank"

# Tokenize sentence
words = word_tokenize(sentence)

# Apply Lesk algorithm
result = lesk(words, 'bank', 'n')

# Display result
print("Sentence:", sentence)
print("Word: bank")

if result:
    print("Meaning:", result.definition())
    print("Synset:", result.name())
else:
    print("No meaning found")
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Documents
documents = [
    "Python is a programming language",
    "Machine learning uses Python",
    "Java is also a programming language"
]

# Search query
query = ["Python programming"]

# Calculate TF-IDF
vectorizer = TfidfVectorizer()
tfidf = vectorizer.fit_transform(documents + query)

# Calculate similarity
similarity = cosine_similarity(tfidf[-1], tfidf[:-1])

# Display document ranking
print("Document Ranking:")

ranking = similarity[0].argsort()[::-1]

for i in ranking:
    print(documents[i], ":", round(similarity[0][i], 2))
```

```python

import spacy

nlp = spacy.load("en_core_web_sm")

text = "Ravi is a student. He studies Python."

doc = nlp(text)

previous_noun = None

print("Reference Resolution:")

for token in doc:
    if token.pos_ == "NOUN" or token.pos_ == "PROPN":
        previous_noun = token.text

    elif token.pos_ == "PRON" and token.text.lower() == "he":
        print(token.text, "refers to", previous_noun)
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Input sentences
sentences = [
    "Python is a programming language",
    "Python is used for machine learning",
    "Machine learning is an important technology"
]

# Calculate similarity
vectorizer = TfidfVectorizer()
matrix = vectorizer.fit_transform(sentences)

similarity = cosine_similarity(matrix)

# Calculate coherence score
score = similarity[0, 1:].mean()

print("Coherence Score:", round(score, 2))

if score > 0.1:
    print("Text is Coherent")
else:
    print("Text is Not Coherent")
```

```python

def dialogue_act(sentence):
    sentence = sentence.lower()

    if sentence.startswith(("hi", "hello", "hey")):
        return "Greeting"
    elif sentence.endswith("?"):
        return "Question"
    elif "thank" in sentence:
        return "Thanking"
    elif sentence.startswith(("bye", "goodbye")):
        return "Farewell"
    else:
        return "Statement"


# Input conversation
sentences = [
    "Hello everyone",
    "How are you?",
    "I am fine",
    "Thank you"
]

# Recognize dialogue acts
for sentence in sentences:
    print(sentence, ":", dialogue_act(sentence))
```
